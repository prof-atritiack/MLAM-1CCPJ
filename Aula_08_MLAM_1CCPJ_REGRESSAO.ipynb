```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
# Regressão Linear
from sklearn.linear_model import LinearRegression
# Árvore de decisão para regressão
from sklearn.tree import DecisionTreeRegressor
# Métricas de avaliação
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
# Carregar os dados utilizando read_csv e delimitador de células ';' e separador decimal '.'
dados = pd.read_csv('https://raw.githubusercontent.com/prof-atritiack/MLAM-1CCPJ/refs/heads/main/datasets_/concrete%2Bcompressive%2Bstrength/Concrete_Data.csv',sep=';')
dados.head()
```

```python
for col in dados.select_dtypes(include=['object']).columns:
    dados[col] = dados[col].str.replace(',', '.', regex=False).astype(float)
dados.info()
```
